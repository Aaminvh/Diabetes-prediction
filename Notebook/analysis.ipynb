### <b>V <span style='color:#16C2D5'>|</span> BMI = 27.32: a hidden missing-value placeholder</b>

The BMI histogram has one very tall spike, and BMI = 27.32 also appeared over and over in the duplicate rows. The checks below test whether 27.32 is a real measurement or a filled-in value, and whether the affected patients differ from the rest.

```python
is_placeholder_raw = df_raw["bmi"] == 27.32
dup_raw = df_raw.duplicated()

print(f"Rows with BMI = 27.32 (raw data): {is_placeholder_raw.sum():,} ({is_placeholder_raw.mean():.1%})")
print(f"Mean BMI of the raw data:         {df_raw['bmi'].mean():.4f}")
print(f"Median BMI of the raw data:       {df_raw['bmi'].median():.2f}")
print(f"\nNext most frequent BMI values:\n{df_raw['bmi'].value_counts().iloc[1:4].to_string()}")
print(f"\nDuplicate rows dropped earlier:   {dup_raw.sum():,}")
print(f"  ...of which have BMI = 27.32:   {(dup_raw & is_placeholder_raw).sum():,}")
```

**📝 Interpretation:** 27.32 occurs 25,495 times in the raw data, while the next most common value occurs only about 100 times. It also equals the **mean of the BMI column to two decimals**, and the median as well. This is the typical sign of **mean imputation**: BMI was missing for roughly a quarter of patients, and the gaps were filled with the column average before the dataset was published. These are missing values, not measurements.

```python
# Are the "duplicates" real re-entries, or distinct patients made identical by the imputed BMI?
# Test: take patients with a *measured* BMI, hide their BMI, and see how often records collide by chance.
non_bmi_cols = [c for c in df_raw.columns if c != "bmi"]
measured_raw = df_raw[~is_placeholder_raw]

chance_rate = np.mean([
    measured_raw.sample(is_placeholder_raw.sum(), random_state=s)[non_bmi_cols].duplicated().mean()
    for s in range(5)
])
print(f"Duplicate rate among BMI = 27.32 rows:                   {df_raw[is_placeholder_raw].duplicated().mean():.1%}")
print(f"Chance collision rate if measured BMI is hidden:         {chance_rate:.1%}")
print(f"Duplicate rate among measured-BMI rows (BMI kept):       {measured_raw.duplicated().mean():.2%}")
```

**📝 Interpretation:** among patients with a measured BMI, true duplicates are almost absent (about 0.03%). If their BMI is hidden, about 12% of their records collide with another patient's purely by chance, which is close to the 15% duplicate rate in the placeholder rows. So most of the dropped "duplicates" are probably distinct patients rather than double entries. This is why the duplicate removal above is flagged for a sensitivity analysis.

```python
# Do patients with an imputed BMI differ from those with a measured BMI?
is_placeholder = df["bmi"] == 27.32
group = is_placeholder.map({False: "Measured BMI", True: "BMI = 27.32"})

comparison = pd.DataFrame({
    "Patients": group.value_counts(),
    "Diabetes (%)": df.groupby(group)["diabetes"].mean() * 100,
    "Mean age": df.groupby(group)["age"].mean(),
    "Hypertension (%)": df.groupby(group)["hypertension"].mean() * 100,
    "Heart disease (%)": df.groupby(group)["heart_disease"].mean() * 100,
    "Smoking 'No Info' (%)": df.groupby(group)["smoking_history"].apply(lambda s: (s == "No Info").mean() * 100),
}).loc[["Measured BMI", "BMI = 27.32"]]
comparison
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5), gridspec_kw={"width_ratios": [2, 1]})

# (a) BMI histogram with the placeholder value shown separately
ax = axes[0]
bins = np.arange(10, 70.5, 0.5)
ax.hist(df.loc[~is_placeholder, "bmi"], bins=bins, color=NEG_COLOR, label="Measured BMI")
ax.hist(df.loc[is_placeholder, "bmi"], bins=bins, color=POS_COLOR, label="BMI = 27.32 (imputed)")
# Cap the y-axis so the measured distribution is readable; the 27.32 bar is clipped, so label its true height
ax.set_ylim(0, 5000)
ax.annotate(f"↑ {is_placeholder.sum():,} patients", (27.32, 5000), xytext=(6, -14), textcoords="offset points",
            ha="left", va="center", fontsize=9, color="#2b2b29")
ax.set_title("BMI distribution: one value accounts for 22.5% of patients")
ax.set_xlabel("BMI (kg/m², shown up to 70)")
ax.set_ylabel("Patients")
ax.legend(frameon=False)

# (b) diabetes prevalence in each group
ax = axes[1]
rates = comparison["Diabetes (%)"]
ax.bar(rates.index, rates.values, color=[NEG_COLOR, POS_COLOR], width=0.55)
label_bars(ax)
ax.axhline(df["diabetes"].mean() * 100, color=MUTED, ls="--", lw=1)
ax.text(1.27, df["diabetes"].mean() * 100 + 0.2, f"overall {df['diabetes'].mean():.1%}", ha="right", va="bottom", fontsize=8, color="#52514e")
ax.set_title("Diabetes prevalence")
ax.set_ylabel("Diabetic (%)")
ax.set_ylim(0, rates.max() * 1.25)

plt.tight_layout()
plt.show()
```

**📝 Interpretation:** the measured BMI values (blue) form a smooth, slightly right-skewed distribution. The imputed rows (orange) all sit in one bar at 27.32, which hides about a fifth of the real spread. The imputed BMI **is not missing at random**: those patients have a lower diabetes prevalence (about 7.0% against 9.4%), less hypertension (5.8% against 8.3%), and much more often have `smoking_history = 'No Info'` (49% against 30%). This pattern suggests patients with less complete records, possibly patients with fewer clinical contacts. Some of them are infants, who were given an adult "overweight" BMI of 27.32, which is clinically impossible.

#### ✅ Decision: keep the rows, treat 27.32 as "imputed", and add a `bmi_imputed` indicator

| Option | Why it was / wasn't chosen |
|---|---|
| Drop the 27.32 rows (complete-case analysis) | ❌ Loses 22.5% of the data. Because the missingness is related to the outcome (7.0% against 9.4% diabetic), the remaining sample would overstate diabetes prevalence and no longer represent the population. |
| Keep 27.32 as if it were measured | ❌ Treats a filled-in number as real. The BMI effect gets diluted toward the mean, and a model cannot tell measured from imputed values. |
| Re-impute BMI with a model (for example from age and gender) | ⏳ Possible later, but it has to be fit inside the training folds to avoid leakage. It adds complexity with little gain, since BMI is a weak predictor next to HbA1c and glucose. |
| **Keep 27.32 (it already is mean imputation) and add a binary `bmi_imputed` flag** | ✅ **Chosen.** No rows are lost. The value is neutral (the mean). The flag records that the value is imputed, so models can learn a separate effect for "BMI not recorded". This is the standard missing-indicator approach. |

**Consequences for the analysis:**
- `bmi_imputed` is created below. It is **not yet used** in the Logistic Regression baseline, so the baseline results are unchanged. It should be added as a feature when the models are compared, especially for the tree models.
- The baseline BMI odds ratio (≈ 1.8 per SD) is probably an **underestimate**, because a quarter of the BMI values carry no information.
- Planned sensitivity analysis: refit the models (a) on measured-BMI rows only, (b) excluding BMI > 80, and (c) without dropping the duplicates, then check whether the conclusions change.

### <b>VI <span style='color:#16C2D5'>|</span> Class imbalance</b>

```python
class_counts = df["diabetes"].value_counts().sort_index()
class_pct = df["diabetes"].value_counts(normalize=True).sort_index() * 100

print(f"Non-diabetic: {class_counts[0]:,} ({class_pct[0]:.1f}%)")
print(f"Diabetic:     {class_counts[1]:,} ({class_pct[1]:.1f}%)")
print(f"Imbalance ratio: {class_counts[0] / class_counts[1]:.1f} : 1")
print(f"Accuracy of a model that always predicts 'non-diabetic': {class_pct[0]:.1f}%")

fig, ax = plt.subplots(figsize=(8, 2.6))
ax.barh([STATUS_LABELS[k] for k in class_counts.index], class_counts.values,
        color=[NEG_COLOR, POS_COLOR], height=0.55)
for i, (n, p) in enumerate(zip(class_counts.values, class_pct.values)):
    ax.text(n + 800, i, f"{n:,}  ({p:.1f}%)", va="center", fontsize=9, color="#2b2b29")
ax.set_xlim(0, class_counts.max() * 1.25)
ax.set_xlabel("Patients")
ax.set_title("Outcome classes: about 1 diabetic patient for every 10 non-diabetic")
ax.grid(axis="y", visible=False)
plt.tight_layout()
plt.show()
```

**📝 Interpretation:** only **8.8%** of patients are diabetic, a ratio of about **10 : 1**. This imbalance shapes how the models should be built and judged:
- **Accuracy is misleading.** A "model" that labels everyone non-diabetic is already 91.2% accurate while missing every diabetic patient. The baseline's 95.7% accuracy should be compared with that 91.2%, not with 50%.
- **Report recall (sensitivity), precision and PR-AUC for the diabetic class.** Clinically, a missed diabetic (false negative) usually costs more than a false alarm, so recall matters most for screening.
- **Use a stratified train/test split** (`stratify=y`) so both sets keep the 8.8% prevalence.
- **Rebalancing:** use class weights, SMOTE or undersampling **only on the training data** (inside an `imbPipeline` during cross-validation), never before the split. Otherwise synthetic or duplicated patients leak into the test set.

# <span style="color:#E888BB; font-size: 1%;">4 | MODEL COMPARISON</span>
<div style="padding: 35px;color:white;margin:10;font-size:170%;text-align:left;display:fill;border-radius:10px;overflow:hidden;background:linear-gradient(90deg, #1a1a2e 0%, #16213e 40%, #0f3460 70%, #533483 100%);"><b>4 | MODEL COMPARISON</b></div>

The section above fitted a first Logistic Regression baseline. This section compares four models under one leak-free setup:

- **Features:** age, BMI, HbA1c, blood glucose, hypertension, heart disease, the `bmi_imputed` flag, gender and smoking status. Smoking status has four levels: never, past, current and **unknown** ("No Info"). "Unknown" is kept as a separate level because the EDA showed it behaves differently from "never".
- **Stratified 80/20 split**, so both sets keep the 8.8% prevalence. The test set is used **once**, at the very end.
- **sklearn `Pipeline`s:** scaling (Logistic Regression only) and one-hot encoding are part of each pipeline, so they are re-fit on the training folds only, inside every CV fold and in the final fit.
- **Class imbalance:** every model up-weights the diabetic class. Logistic Regression, Decision Tree and Random Forest use `class_weight="balanced"`, and XGBoost uses `scale_pos_weight`. This keeps recall and precision at the default 0.5 threshold comparable across models.
- **Evaluation:** 5-fold stratified CV on the training set (ROC-AUC, PR-AUC, recall, precision, F1), then the held-out test set.

```python
# Modelling frame. Smoking history keeps "No Info" as its own level ("unknown"): the EDA showed
# these patients have a much lower diabetes rate than never-smokers, so folding them in loses signal.
smoking_map = {"never": "never", "No Info": "unknown", "current": "current",
               "former": "past", "ever": "past", "not current": "past"}

model_df = df[["age", "gender", "hypertension", "heart_disease", "bmi", "bmi_imputed",
               "HbA1c_level", "blood_glucose_level", "diabetes"]].copy()
model_df["smoking_status"] = df_raw.loc[df.index, "smoking_history"].map(smoking_map)

NUM_FEATURES = ["age", "bmi", "HbA1c_level", "blood_glucose_level"]
BIN_FEATURES = ["hypertension", "heart_disease", "bmi_imputed"]
CAT_FEATURES = ["gender", "smoking_status"]
FEATURES = NUM_FEATURES + BIN_FEATURES + CAT_FEATURES

X_all = model_df[FEATURES]
y_all = model_df["diabetes"]

# Stratified 80/20 split: both sets keep the 8.8% prevalence
X_tr, X_te, y_tr, y_te = train_test_split(X_all, y_all, test_size=0.2, stratify=y_all, random_state=42)
print(f"Train: {len(X_tr):,} rows ({y_tr.mean():.2%} diabetic) | Test: {len(X_te):,} rows ({y_te.mean():.2%} diabetic)")
```

**📝 Note:** both sets keep exactly 8.82% diabetic patients, so the stratification worked. This is a new split, so the results below can't be compared row-for-row with the first baseline in section 3. That baseline used an unstratified split and different features.

### <b>I <span style='color:#16C2D5'>|</span> Pipelines</b>

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (roc_auc_score, average_precision_score, recall_score, precision_score,
                             f1_score, roc_curve, precision_recall_curve)

RANDOM_STATE = 42
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
SCORING = {"ROC-AUC": "roc_auc", "PR-AUC": "average_precision", "Recall": "recall",
           "Precision": "precision", "F1": "f1"}

# Preprocessing lives inside each pipeline, so in every CV fold (and in the final fit)
# the scaler and encoder are fit on training rows only.
linear_prep = ColumnTransformer([
    ("num", StandardScaler(), NUM_FEATURES),
    ("bin", "passthrough", BIN_FEATURES),
    ("cat", OneHotEncoder(drop="first", handle_unknown="ignore"), CAT_FEATURES),
])
tree_prep = ColumnTransformer([      # trees do not need scaling
    ("num", "passthrough", NUM_FEATURES + BIN_FEATURES),
    ("cat", OneHotEncoder(handle_unknown="ignore"), CAT_FEATURES),
])

# Every model re-weights the minority class, so recall / precision at the default 0.5
# threshold are comparable across models.
neg_pos_ratio = (y_tr == 0).sum() / (y_tr == 1).sum()

pipelines = {
    "Logistic Regression": Pipeline([("prep", linear_prep),
        ("model", LogisticRegression(class_weight="balanced", max_iter=1000))]),
    "Decision Tree": Pipeline([("prep", tree_prep),
        ("model", DecisionTreeClassifier(class_weight="balanced", random_state=RANDOM_STATE))]),
    "Random Forest": Pipeline([("prep", tree_prep),
        ("model", RandomForestClassifier(n_estimators=200, class_weight="balanced", n_jobs=4,
                                         random_state=RANDOM_STATE))]),
    "XGBoost": Pipeline([("prep", tree_prep),
        ("model", XGBClassifier(scale_pos_weight=neg_pos_ratio, eval_metric="logloss", tree_method="hist",
                                n_jobs=-1, random_state=RANDOM_STATE))]),
}
```

### <b>II <span style='color:#16C2D5'>|</span> Light hyperparameter tuning (tree models)</b>

```python
# Light tuning of the tree models: small grids, 5-fold stratified CV on the training set,
# selecting on PR-AUC (the most informative single metric for a rare outcome).
param_grids = {
    "Decision Tree": {"model__max_depth": [4, 8, 12, 16], "model__min_samples_leaf": [20, 100, 300]},
    # Random Forest depth is capped (no fully grown trees) to keep memory use modest
    "Random Forest": {"model__max_depth": [12, 20], "model__min_samples_leaf": [5, 20]},
    "XGBoost": {"model__n_estimators": [200, 400], "model__max_depth": [3, 5],
                "model__learning_rate": [0.05, 0.1]},
}

best_params = {}
for name, grid in param_grids.items():
    search = GridSearchCV(pipelines[name], grid, scoring="average_precision", cv=cv)
    search.fit(X_tr, y_tr)
    pipelines[name] = search.best_estimator_
    best_params[name] = {k.replace("model__", ""): v for k, v in search.best_params_.items()}
    print(f"{name:<14} best PR-AUC {search.best_score_:.4f}  {best_params[name]}")
```

**📝 Interpretation:** the tuning picked **regularised** trees. The Decision Tree uses large leaves (at least 100 patients each), the Random Forest uses depth 20 with at least 20 patients per leaf, and XGBoost prefers **shallow trees (depth 3)** with a slow learning rate (0.05, 400 rounds). Few, coarse splits are enough, which fits the threshold-like structure found in the EDA. The grids are deliberately small. The Decision Tree and Random Forest picked values at the edge of their grids, so somewhat better settings may exist. The CV scores used to choose the settings are also slightly optimistic, which is why the untouched test set is checked below.

### <b>III <span style='color:#16C2D5'>|</span> Cross-validated comparison</b>

```python
cv_rows = {}
for name, pipe in pipelines.items():
    scores = cross_validate(pipe, X_tr, y_tr, cv=cv, scoring=SCORING)
    cv_rows[name] = {m: f"{scores['test_' + m].mean():.3f} ± {scores['test_' + m].std():.3f}" for m in SCORING}

cv_table = pd.DataFrame(cv_rows).T
cv_table.index.name = "5-fold CV (train set), mean ± SD"
cv_table
```

**📝 Interpretation:**
- **All four models discriminate very well** (ROC-AUC 0.96–0.98). With 91% easy negatives, ROC-AUC flatters every model. **PR-AUC** separates them better: Logistic Regression 0.82 < Decision Tree 0.87 < Random Forest 0.88 < **XGBoost 0.89**.
- The results are **very stable across folds** (SD ≤ 0.013), so the ranking is not down to luck in the split.
- Because every model up-weights diabetics, **recall is high (0.88–0.92)** and **precision is modest (0.43–0.52)**. At the default threshold, roughly one in two patients flagged as diabetic actually has diabetes. XGBoost has the highest recall. The Random Forest has the best precision and F1.

**📝 Interpretation:** the test-set results agree with the CV estimates to within about 0.01, so the models are not overfitted to the tuning folds. The test set has 1,696 diabetic patients:
- **XGBoost misses 140** of them (recall 0.92). Logistic Regression misses 209 and the Random Forest 199.
- For comparison, the **unweighted** baseline in section 3 missed **655** (recall 0.61). Most of that gain comes from **class weighting**, not from the choice of algorithm. The cost is far more **false alarms** (about 1,400–2,000, against 171 for the unweighted baseline).
- In screening, a false alarm usually just means a confirmatory HbA1c or fasting glucose test, so this trade-off is often acceptable. The right balance is a clinical decision, made by choosing the threshold (next plot).

# <span style="color:#E888BB; font-size: 1%;">5 | CALIBRATION, THRESHOLD AND INTERPRETATION</span>

This section takes the best model from section 4 (**tuned, class-weighted XGBoost**) and turns it into something usable for screening:

1. **Calibration.** Class weighting inflates the predicted probabilities. Sigmoid (Platt) and isotonic calibration are compared with `CalibratedClassifierCV`.
2. **Decision threshold.** Chosen from **out-of-fold predictions on the training set only**. The test set is used once, to report how the chosen thresholds perform.
3. **SHAP feature importance**, to see what the model has learnt.
4. **Sensitivity analyses** for the data-quality decisions made in section 1.

To keep memory low, all cross-validation runs sequentially (no `n_jobs`) and SHAP values are computed on a 5,000-row sample.

### <b>I <span style='color:#16C2D5'>|</span> Probability calibration</b>

`CalibratedClassifierCV` learns a mapping from raw model scores to probabilities on rows the boosted trees did not see (inner 3-fold CV). **Sigmoid** fits a logistic curve, which has 2 parameters and is robust with little data. **Isotonic** fits a free-form monotone step function, which is more flexible and needs more data. With about 6,800 diabetics in the training set, isotonic has enough data.

To compare them using training data only, each candidate is wrapped in the outer 5-fold CV from section 4 (`cross_val_predict`). Every training row then gets an **out-of-fold (OOF)** probability. An unweighted XGBoost is included as a reference. The **Brier score** is the mean squared error of the predicted probability (lower is better). Calibration does not change the ranking much, so PR-AUC and ROC-AUC should barely move.

```python
from sklearn.base import clone
from sklearn.calibration import CalibratedClassifierCV, calibration_curve
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import brier_score_loss, log_loss

# Tuned, class-weighted XGBoost pipeline from section 4
xgb_best = pipelines["XGBoost"]
inner_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)

calib_candidates = {
    "Weighted, uncalibrated": clone(xgb_best),
    "Weighted + sigmoid": CalibratedClassifierCV(clone(xgb_best), method="sigmoid", cv=inner_cv),
    "Weighted + isotonic": CalibratedClassifierCV(clone(xgb_best), method="isotonic", cv=inner_cv),
    "Unweighted (reference)": clone(xgb_best).set_params(model__scale_pos_weight=1),
}

# Out-of-fold probabilities for every training row (outer 5-fold CV, run sequentially)
oof_proba = {name: cross_val_predict(est, X_tr, y_tr, cv=cv, method="predict_proba")[:, 1]
             for name, est in calib_candidates.items()}

calib_table = pd.DataFrame({
    name: {"Brier score": brier_score_loss(y_tr, p), "Log loss": log_loss(y_tr, p),
           "Mean predicted risk": p.mean(), "PR-AUC": average_precision_score(y_tr, p),
           "ROC-AUC": roc_auc_score(y_tr, p)}
    for name, p in oof_proba.items()}).T
calib_table.index.name = f"Training-set OOF (observed prevalence {y_tr.mean():.3f})"

# Calibration method with the lower OOF Brier score
best_calib = min(["sigmoid", "isotonic"], key=lambda m: calib_table.loc[f"Weighted + {m}", "Brier score"])
print(f"Chosen calibration method: {best_calib}")
calib_table.style.format("{:.4f}")
```

```python
CALIB_COLORS = {"Weighted, uncalibrated": "#2a78d6", "Weighted + sigmoid": "#eb6834",
                "Weighted + isotonic": "#1baf7a"}

fig, axes = plt.subplots(1, 2, figsize=(13, 5), gridspec_kw={"width_ratios": [1.15, 1]})
ax = axes[0]
ax.plot([0, 1], [0, 1], color=MUTED, ls="--", lw=1, label="Perfect calibration")
for name, color in CALIB_COLORS.items():
    frac_pos, mean_pred = calibration_curve(y_tr, oof_proba[name], n_bins=10, strategy="uniform")
    ax.plot(mean_pred, frac_pos, "-o", color=color, lw=2, ms=6, markeredgecolor="white",
            label=f"{name} (Brier {calib_table.loc[name, 'Brier score']:.4f})")
ax.set(title="Calibration curve (training-set OOF predictions)", xlabel="Mean predicted probability",
       ylabel="Observed fraction diabetic", xlim=(0, 1), ylim=(0, 1.02))
ax.legend(frameon=False, loc="upper left", fontsize=9)

ax = axes[1]
bins = np.linspace(0, 1, 41)
for name, color in CALIB_COLORS.items():
    ax.hist(oof_proba[name], bins=bins, histtype="step", lw=2, color=color, label=name)
ax.set_yscale("log")
ax.set(title="Distribution of predicted probabilities", xlabel="Predicted probability",
       ylabel="Number of patients (log scale)")
ax.legend(frameon=False, fontsize=9)
for a in axes:
    a.spines[["top", "right"]].set_visible(False)
    a.grid(True, color="#e6e5e1", linewidth=0.8)
plt.tight_layout()
plt.show()
```

**📝 Interpretation:** - The **class-weighted model badly overstates risk.** Its mean predicted risk is 18.3% against an observed 8.8%, and the blue curve sits far below the diagonal. For example, patients given a 50% risk are diabetic only about 10% of the time. This is the expected side effect of `scale_pos_weight` ≈ 10, which trains the model as if diabetics were 10 times more common. Brier score 0.059.
- **Sigmoid** calibration corrects the average (8.83%), but its S-shape cannot follow this distortion and it still overstates risk in the middle range (Brier 0.026).
- **Isotonic** calibration follows the diagonal closely up to about 0.65 and has the lowest Brier score (**0.0232**), so it is chosen. Above 0.7 the curve is noisy because the bins are nearly empty. The histogram shows that calibrated predictions are pushed to near 0 or near 1 (the patients above the HbA1c/glucose cut-offs, who are always diabetic), so only a few dozen patients fall in between.
- The **unweighted** XGBoost is just as well calibrated (Brier 0.0232) and ranks patients just as well (same PR-AUC). Class weighting therefore did not improve discrimination. Its only effect was to move where the default 0.5 cut-off falls. Calibrated probabilities with an explicitly chosen threshold (next part) are a cleaner way to get high recall.
- PR-AUC (0.890) and ROC-AUC (0.980) are unchanged, so calibration changes the probability scale, not the ranking of patients.

```python
# Final calibrated model: the chosen method, calibrated with the same 5-fold CV on the full training set.
# The test set is touched once here, to confirm the calibration holds on unseen patients.
xgb_calibrated = CalibratedClassifierCV(clone(xgb_best), method=best_calib, cv=cv).fit(X_tr, y_tr)
p_te_cal = xgb_calibrated.predict_proba(X_te)[:, 1]
p_te_raw = test_proba["XGBoost"]     # weighted, uncalibrated (section 4)

pd.DataFrame({
    "Weighted, uncalibrated": {"Brier score": brier_score_loss(y_te, p_te_raw), "Mean predicted risk": p_te_raw.mean(),
                               "PR-AUC": average_precision_score(y_te, p_te_raw)},
    f"Weighted + {best_calib}": {"Brier score": brier_score_loss(y_te, p_te_cal), "Mean predicted risk": p_te_cal.mean(),
                                 "PR-AUC": average_precision_score(y_te, p_te_cal)},
}).T.rename_axis(f"Held-out test set (prevalence {y_te.mean():.3f})").style.format("{:.4f}")
```

**📝 Note:** on the held-out test set, isotonic calibration cuts the Brier score from 0.061 to **0.023**, and the mean predicted risk (9.0%) now matches the observed prevalence (8.8%). PR-AUC is unchanged (0.882), so calibration cost no discrimination. The calibrated probabilities can be read as actual risks, at least in a population with this prevalence.

### <b>II <span style='color:#16C2D5'>|</span> Choosing the decision threshold</b>

The threshold is chosen from the **calibrated out-of-fold probabilities on the training set**, never from the test set. Four candidates:

- **Recall ≈ 0.95, 0.90 and 0.85**: the highest threshold that still reaches each target recall (sensitivity) in the OOF predictions.
- **Best F1**: the threshold that maximises F1, which weights precision and recall equally.

Because the probabilities are calibrated, a threshold can be read as a risk: "flag everyone with an estimated diabetes risk of at least *t*". The candidates are then applied **once** to the held-out test set.

```python
def operating_point(y_true, proba, threshold):
    """Confusion-matrix counts and screening metrics at one threshold."""
    pred = proba >= threshold
    y_true = np.asarray(y_true) == 1
    tp, fp = (pred & y_true).sum(), (pred & ~y_true).sum()
    fn, tn = (~pred & y_true).sum(), (~pred & ~y_true).sum()
    return {"Threshold": threshold, "Recall": tp / (tp + fn), "Precision": tp / (tp + fp),
            "Specificity": tn / (tn + fp), "F1": 2 * tp / (2 * tp + fp + fn),
            "Flagged per 1,000": 1000 * pred.mean(), "Flagged per case found": (tp + fp) / tp,
            "TP": tp, "FP": fp, "FN": fn, "TN": tn}


p_oof = oof_proba[f"Weighted + {best_calib}"]
prec_oof, rec_oof, thr_oof = precision_recall_curve(y_tr, p_oof)
prec_oof, rec_oof = prec_oof[:-1], rec_oof[:-1]          # align with thr_oof


def threshold_for_recall(target):
    # recall falls as the threshold rises: take the highest threshold that still reaches the target
    return thr_oof[np.where(rec_oof >= target)[0].max()]


f1_oof = 2 * prec_oof * rec_oof / np.clip(prec_oof + rec_oof, 1e-12, None)
candidates = {"Recall ≈ 0.95": threshold_for_recall(0.95), "Recall ≈ 0.90": threshold_for_recall(0.90),
              "Recall ≈ 0.85": threshold_for_recall(0.85), "Best F1": thr_oof[np.argmax(f1_oof)]}

METRIC_FMT = {"Threshold": "{:.3f}", "Recall": "{:.3f}", "Precision": "{:.3f}", "Specificity": "{:.3f}",
              "F1": "{:.3f}", "Flagged per 1,000": "{:.0f}", "Flagged per case found": "{:.2f}"}
COUNT_COLS = ["TP", "FP", "FN", "TN"]

thr_train = pd.DataFrame({k: operating_point(y_tr, p_oof, t) for k, t in candidates.items()}).T
thr_train[COUNT_COLS] = thr_train[COUNT_COLS].astype(int)
thr_train.index.name = "Training set, OOF (chosen here)"
thr_train.style.format(METRIC_FMT)
```

```python
# Apply the thresholds chosen above, unchanged, to the held-out test set.
# The default 0.5 cut-off of the uncalibrated weighted model (section 4) is shown for reference.
test_points = {"Default 0.5 (uncalibrated)": operating_point(y_te, p_te_raw, 0.5)}
test_points.update({k: operating_point(y_te, p_te_cal, t) for k, t in candidates.items()})
thr_test = pd.DataFrame(test_points).T
thr_test[COUNT_COLS] = thr_test[COUNT_COLS].astype(int)
thr_test.index.name = "Held-out test set (applied once)"
thr_test.style.format(METRIC_FMT)
```

```python
from matplotlib.colors import LinearSegmentedColormap

fig, axes = plt.subplots(1, 4, figsize=(16, 3.9))
cm_cmap = LinearSegmentedColormap.from_list("cm", ["#f4f3ef", "#2a78d6"])
for ax, (name, row) in zip(axes, thr_test.loc[list(candidates)].iterrows()):
    cm = np.array([[row["TN"], row["FP"]], [row["FN"], row["TP"]]], dtype=int)
    # colour by row share, so the rare diabetic row stays readable next to 17,000 negatives
    ax.imshow(cm / cm.sum(axis=1, keepdims=True), cmap=cm_cmap, vmin=0, vmax=1)
    for i in range(2):
        for j in range(2):
            share = cm[i, j] / cm[i].sum()
            ax.text(j, i, f"{cm[i, j]:,}\n({share:.1%})", ha="center", va="center", fontsize=10,
                    color="white" if share > 0.6 else "#2b2b29")
    ax.set_xticks([0, 1], ["Not flagged", "Flagged"])
    ax.set_yticks([0, 1], ["Non-diabetic", "Diabetic"])
    ax.set_title(f"{name}\nthreshold {row['Threshold']:.3f}", fontsize=11)
    ax.grid(False)
    ax.spines[:].set_visible(False)
axes[0].set_ylabel("True status")
fig.suptitle("Test-set confusion matrices at the CV-chosen thresholds (cell colour = share of true-status row)",
             fontsize=12, fontweight="bold", y=1.03)
plt.tight_layout()
plt.show()
```

```python
prec_te, rec_te, _ = precision_recall_curve(y_te, p_te_cal)
fig, ax = plt.subplots(figsize=(8.5, 5.5))
ax.plot(rec_oof, prec_oof, color=MUTED, lw=2, label="Training set, OOF (thresholds chosen here)")
ax.plot(rec_te, prec_te, color="#2a78d6", lw=2, label="Held-out test set")
POINT_COLORS = ["#eb6834", "#1baf7a", "#eda100", "#533483"]
for color, (name, row) in zip(POINT_COLORS, thr_test.loc[list(candidates)].iterrows()):
    ax.plot(row["Recall"], row["Precision"], "o", ms=10, color=color, markeredgecolor="white", markeredgewidth=1.5,
            label=f"{name}: t = {row['Threshold']:.3f} (test recall {row['Recall']:.2f}, precision {row['Precision']:.2f})")
ax.axhline(y_te.mean(), color=MUTED, ls="--", lw=1)
ax.set(title="Precision–recall trade-off, calibrated XGBoost", xlabel="Recall (sensitivity)",
       ylabel="Precision (PPV)", xlim=(0.5, 1.0), ylim=(0, 1.02))
ax.spines[["top", "right"]].set_visible(False)
ax.grid(True, color="#e6e5e1", linewidth=0.8)
ax.legend(frameon=False, loc="lower left", fontsize=8.5)
plt.tight_layout()
plt.show()
```

**📝 Interpretation and recommendation:** - **The thresholds chosen on the training set transfer well to the test set.** Test recall lands 0.006–0.013 below each target (0.944, 0.891, 0.837). A slight shortfall is expected, because each threshold sits right at the edge of its target. The test precision–recall curve follows the OOF curve closely.
- Because the probabilities are calibrated, the thresholds are **risks**. t = 0.071 means "flag anyone with an estimated risk of 7% or more", which is just below the 8.8% prevalence.
- **Best F1 (t = 0.37) is not a screening threshold.** It misses 480 of 1,696 diabetics (28%). It flags essentially only the patients above the HbA1c/glucose cut-offs, where precision is close to 1.
- **Marginal cost** on the test set (19,226 patients): moving from recall 0.85 to 0.90 finds 91 more diabetics for 542 more false alarms, about 6 extra confirmatory tests per extra case found. Moving from 0.90 to 0.95 finds 90 more for 858 more false alarms, about 9.5 per extra case.
- Compared with the default 0.5 cut-off of the uncalibrated model (recall 0.917, 1,769 false alarms), the recall-0.95 threshold finds 45 more diabetics for 484 more false alarms.

**✅ Recommendation for screening: threshold ≈ 0.07 (target recall 0.95).** In screening, a missed case means undiagnosed diabetes and a delayed start of treatment, while a false alarm costs one cheap, non-invasive confirmatory test (HbA1c or fasting glucose). On the test set this threshold:
- catches **94.4%** of diabetics and misses 95 of 1,696;
- flags 20% of screened patients, about **2.4 flagged per true case**;
- gives a negative predictive value of **99.4%** (15,277 of 15,372 unflagged patients are non-diabetic), so a negative result is reassuring.

If confirmatory-testing capacity is limited, **t ≈ 0.11 (recall ≈ 0.90)** is the fallback. It flags 15% of patients instead of 20% (1.9 per case found). The threshold should be re-chosen for any population with a different prevalence. There is also an important caveat: the decisive inputs here, HbA1c and glucose, are themselves the confirmatory tests (see Limitations).

### <b>III <span style='color:#16C2D5'>|</span> SHAP feature importance</b>

SHAP values split each patient's prediction into additive contributions, one per feature: *baseline + Σ contributions = model output*. They are computed with XGBoost's built-in exact **TreeSHAP** (`pred_contribs=True`), so no extra package is needed. The values are on the **log-odds scale of the weighted model**. Calibration is a monotone transformation applied afterwards, so it does not change which features drive the ranking. The one-hot columns of gender and smoking status are summed back into a single feature each (SHAP values are additive), and the values are computed on a random sample of 5,000 test-set patients.

```python
import xgboost

shap_sample = X_te.sample(5000, random_state=RANDOM_STATE)
prep_fitted = xgb_best.named_steps["prep"]
X_shap = prep_fitted.transform(shap_sample)
X_shap = X_shap.toarray() if hasattr(X_shap, "toarray") else X_shap
enc_names = [n.split("__", 1)[1] for n in prep_fitted.get_feature_names_out()]

contribs = xgb_best.named_steps["model"].get_booster().predict(xgboost.DMatrix(X_shap), pred_contribs=True)
shap_enc = pd.DataFrame(contribs[:, :-1], columns=enc_names, index=shap_sample.index)   # last column = bias

# Sum the one-hot columns back into their original feature
shap_df = pd.DataFrame({f: shap_enc[[c for c in enc_names if c == f or c.startswith(f + "_")]].sum(axis=1)
                        for f in FEATURES})
shap_importance = shap_df.abs().mean().sort_values()
del contribs, shap_enc, X_shap

fig, axes = plt.subplots(1, 2, figsize=(15, 5.2), gridspec_kw={"width_ratios": [1, 1.5]})
ax = axes[0]
ax.barh(shap_importance.index, shap_importance.values, color="#2a78d6", height=0.6)
for y_pos, v in enumerate(shap_importance.values):
    ax.text(v, y_pos, f" {v:.2f}", va="center", fontsize=9, color="#2b2b29")
ax.set(title="Mean |SHAP value| (log-odds)", xlabel="Average impact on model output")
ax.grid(axis="y", visible=False)

# Beeswarm-style view: one dot per patient, coloured by the feature's value (percentile within the sample)
ax = axes[1]
rng = np.random.default_rng(RANDOM_STATE)
order = shap_importance.index
for y_pos, f in enumerate(order):
    vals = shap_df[f].to_numpy()
    if f in CAT_FEATURES:
        colors = MUTED
    elif f in BIN_FEATURES:
        colors = shap_sample[f].to_numpy()                 # 0/1 → low/high
    else:
        colors = shap_sample[f].rank(pct=True).to_numpy()  # percentile, robust to outliers
    ax.scatter(vals, y_pos + rng.uniform(-0.3, 0.3, len(vals)), c=colors, cmap="viridis", s=4, alpha=0.5,
               vmin=0, vmax=1, rasterized=True)
ax.axvline(0, color="#52514e", lw=0.8)
ax.set_yticks(range(len(order)), order)
ax.set(title="SHAP value per patient (colour = feature value, low → high)", xlabel="SHAP value (log-odds)")
ax.grid(axis="y", visible=False)
sm = plt.cm.ScalarMappable(cmap="viridis", norm=plt.Normalize(0, 1))
cbar = fig.colorbar(sm, ax=ax, pad=0.01, ticks=[0, 1])
cbar.ax.set_yticklabels(["low", "high"])
cbar.set_label("Feature value (percentile; binary 0/1; grey = categorical)", fontsize=9)
for a in axes:
    a.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
```

- Patients with the **imputed BMI (27.32)** get a contribution close to zero. Mean imputation makes BMI neutral for them, which is why the `bmi_imputed` flag adds nothing.

### <b>IV <span style='color:#16C2D5'>|</span> Sensitivity analyses</b>

Section 1 made three data-quality decisions. Each is reversed here to see whether the conclusions depend on it:

| Variant | What changes | Why it matters |
|---|---|---|
| **Duplicates kept** | the 3,854 dropped duplicate rows are put back | 3,829 of them have the imputed BMI 27.32, so they are probably distinct patients rather than re-entries |
| **Measured BMI only** | rows with BMI = 27.32 are removed (`bmi_imputed` is then constant) | checks that the imputed BMI and its flag do not drive the results |
| **BMI > 80 excluded** | physiologically implausible BMI values are removed | checks the influence of probable data-entry errors |

Every variant is scored with the **same tuned XGBoost settings** (not re-tuned). `scale_pos_weight` is recomputed for each variant's prevalence, and each is scored with 5-fold stratified CV on the **whole variant dataset**. Using whole datasets keeps the comparison simple, so these numbers sit next to, not in place of, the train/test results above. With duplicates kept, identical rows could fall into both the training and validation folds and inflate the scores. That variant is therefore scored with **grouped CV**, which keeps identical rows in the same fold. A naive CV row is shown too, to measure that leakage.

```python
from sklearn.model_selection import GroupKFold


def build_model_frame(source):
    """Rebuild the section-4 modelling frame from (a subset of) the raw data."""
    out = source[source["gender"] != "Other"].copy()
    out["bmi_imputed"] = (out["bmi"] == 27.32).astype(int)
    out["smoking_status"] = out["smoking_history"].map(smoking_map)
    return out


primary = build_model_frame(df_raw.drop_duplicates())
with_dups = build_model_frame(df_raw)
assert len(primary) == len(model_df)          # same rows as the main analysis

dup_groups = pd.util.hash_pandas_object(with_dups.drop(columns=["bmi_imputed", "smoking_status"]), index=False)
group_cv = GroupKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

variants = {
    "Primary analysis (duplicates dropped)": (primary, cv, None),
    "Duplicates kept (grouped CV)": (with_dups, group_cv, dup_groups),
    "Duplicates kept (naive CV, leaky)": (with_dups, cv, None),
    "Measured BMI only": (primary[primary["bmi_imputed"] == 0], cv, None),
    "BMI > 80 excluded": (primary[primary["bmi"] <= 80], cv, None),
}

sens_rows = {}
for name, (frame, splitter, groups) in variants.items():
    Xv, yv = frame[FEATURES], frame["diabetes"]
    est = clone(xgb_best).set_params(model__scale_pos_weight=(yv == 0).sum() / (yv == 1).sum())
    scores = cross_validate(est, Xv, yv, cv=splitter, groups=groups, scoring=SCORING)
    sens_rows[name] = {"Rows": len(frame), "Diabetic %": 100 * yv.mean(),
                       **{m: scores["test_" + m].mean() for m in SCORING}}

sens_table = pd.DataFrame(sens_rows).T.astype({"Rows": int})
sens_table.index.name = "XGBoost, 5-fold CV (threshold 0.5)"
del with_dups, dup_groups
sens_table.style.format({"Rows": "{:,}", "Diabetic %": "{:.2f}",
                         **{m: "{:.3f}" for m in SCORING}})
```

**📝 Interpretation:** - **None of the three data-quality decisions changes the conclusions.** In every variant, PR-AUC stays between 0.888 and 0.896, ROC-AUC between 0.979 and 0.980, and recall at 0.5 between 0.922 and 0.927. These differences are within the fold-to-fold SD of about 0.01 seen in section 4.
- **Duplicates kept:** 99,982 rows, with prevalence falling to 8.50% (the duplicate rows are mostly non-diabetic). Performance is practically identical to the primary analysis. Grouped and naive CV agree to three decimals, so the duplicates would not have caused noticeable leakage either. Whether duplicates are dropped does not matter for this model.
- **Measured BMI only:** slightly higher PR-AUC (0.896) and precision (0.496). Part of this is prevalence, not a better model: these patients are more often diabetic (9.36% against 8.82%), and PR-AUC and precision rise with prevalence. Since BMI is a minor predictor (SHAP), the imputed values cost little.
- **BMI > 80 excluded:** only 9 patients are removed, and nothing changes.
- These checks use CV on each whole dataset, with no separate test set. They show that the results are **robust to the choices**. They do not provide new performance estimates.

# <span style="color:#E888BB; font-size: 1%;">6 | NON-LABORATORY SCREENING MODEL</span>
<div style="padding: 35px;color:white;margin:10;font-size:170%;text-align:left;display:fill;border-radius:10px;overflow:hidden;background:linear-gradient(90deg, #1a1a2e 0%, #16213e 40%, #0f3460 70%, #533483 100%);"><b>6 | NON-LABORATORY SCREENING MODEL</b></div>

Sections 4 and 5 use HbA1c and blood glucose. These are the tests that **define** diabetes (ADA/WHO criteria), and in this dataset the labels follow them almost like a rule. That model answers "does this patient, whose lab results we already have, have diabetes?". The more useful screening question comes **before** any blood test: **"who should be sent for an HbA1c or glucose test?"**

This section answers that question with **non-laboratory predictors only**: age, BMI (with the `bmi_imputed` flag), hypertension, heart disease, gender and smoking status. Established questionnaire scores such as the ADA risk test and FINDRISC use the same kind of information. The setup reuses sections 4 and 5:

- **Same stratified 80/20 split and same patients** (`X_tr`/`X_te`, with HbA1c and glucose dropped), so the results compare directly with the full model.
- **Class-weighted Logistic Regression and XGBoost** in the same pipelines. XGBoost gets the same light `GridSearchCV` on PR-AUC.
- **Calibration** (sigmoid against isotonic) and **thresholds** chosen from training-set out-of-fold predictions, then applied **once** to the test set.

### <b>I <span style='color:#16C2D5'>|</span> Models without HbA1c and glucose</b>

```python
# Non-laboratory feature set: everything except HbA1c and blood glucose
NL_NUM = ["age", "bmi"]
NL_BIN = ["hypertension", "heart_disease", "bmi_imputed"]
NL_CAT = ["gender", "smoking_status"]
NL_FEATURES = NL_NUM + NL_BIN + NL_CAT

# Same patients and same split as sections 4 and 5, only fewer columns
Xn_tr, Xn_te = X_tr[NL_FEATURES], X_te[NL_FEATURES]

nl_linear_prep = ColumnTransformer([
    ("num", StandardScaler(), NL_NUM),
    ("bin", "passthrough", NL_BIN),
    ("cat", OneHotEncoder(drop="first", handle_unknown="ignore"), NL_CAT),
])
nl_tree_prep = ColumnTransformer([
    ("num", "passthrough", NL_NUM + NL_BIN),
    ("cat", OneHotEncoder(handle_unknown="ignore"), NL_CAT),
])

nl_pipelines = {
    "Logistic Regression": Pipeline([("prep", nl_linear_prep),
        ("model", LogisticRegression(class_weight="balanced", max_iter=1000))]),
    "XGBoost": Pipeline([("prep", nl_tree_prep),
        ("model", XGBClassifier(scale_pos_weight=neg_pos_ratio, eval_metric="logloss", tree_method="hist",
                                n_jobs=-1, random_state=RANDOM_STATE))]),
}

# Same light XGBoost grid as section 4, selected on PR-AUC (sequential CV to keep memory low)
nl_search = GridSearchCV(nl_pipelines["XGBoost"], param_grids["XGBoost"], scoring="average_precision", cv=cv)
nl_search.fit(Xn_tr, y_tr)
nl_pipelines["XGBoost"] = nl_search.best_estimator_
print("XGBoost best:", {k.replace("model__", ""): v for k, v in nl_search.best_params_.items()})

nl_cv_rows = {}
for name, pipe in nl_pipelines.items():
    scores = cross_validate(pipe, Xn_tr, y_tr, cv=cv, scoring=SCORING)
    nl_cv_rows[f"Non-lab {name}"] = {m: f"{scores['test_' + m].mean():.3f} ± {scores['test_' + m].std():.3f}"
                                     for m in SCORING}
nl_cv_rows["Full XGBoost, with labs (section 4)"] = cv_table.loc["XGBoost"].to_dict()

nl_cv_table = pd.DataFrame(nl_cv_rows).T
nl_cv_table.index.name = "5-fold CV (train set, threshold 0.5), mean ± SD"
nl_cv_table
```

**📝 Interpretation:**
- **Without HbA1c and glucose, discrimination falls sharply but stays useful.** ROC-AUC drops from 0.98 to about **0.83**, and PR-AUC from 0.89 to about **0.32**, against a no-skill PR-AUC of 0.088 (the prevalence). So the non-laboratory model ranks patients about 3.7 times better than chance on PR-AUC. Its performance is in the range reported for questionnaire risk scores, and it is a more realistic measure of what can be predicted **before** a blood test.
- **Logistic Regression is almost as good as XGBoost** (PR-AUC 0.316 against 0.328, ROC-AUC 0.832 against 0.835). With only smooth predictors such as age and BMI there are no sharp cut-offs for the trees to exploit, so the advantage of the trees from section 4 largely disappears. Tuning again picked the shallowest, slowest XGBoost (depth 3, learning rate 0.05, 200 trees).
- At the default 0.5 threshold, the class-weighted models find about 80% of diabetics, but **only about 1 in 5 flagged patients is diabetic** (precision 0.21–0.22), against about 1 in 2 with labs. The variation across folds is small (SD ≤ 0.011).

### <b>II <span style='color:#16C2D5'>|</span> Odds ratios of the non-laboratory predictors</b>

An **unweighted** Logistic Regression is fitted on the training set for interpretation, as in section 3. Class weighting mainly shifts the intercept, so it is left out here to keep the odds ratios on the natural scale. To make them clinically readable, the odds ratios for the continuous predictors are given **per 10 years of age** and **per 5 kg/m² of BMI**, not per standard deviation.

```python
nl_lr_unweighted = clone(nl_pipelines["Logistic Regression"]).set_params(model__class_weight=None)
nl_lr_unweighted.fit(Xn_tr, y_tr)

prep_lr = nl_lr_unweighted.named_steps["prep"]
coefs = pd.Series(nl_lr_unweighted.named_steps["model"].coef_[0],
                  index=[n.split("__", 1)[1] for n in prep_lr.get_feature_names_out()])

# Convert the per-SD coefficients of the scaled continuous predictors into clinical units
sd = pd.Series(prep_lr.named_transformers_["num"].scale_, index=NL_NUM)
unit = {"age": 10, "bmi": 5}
labels = {"age": "Age (per 10 years)", "bmi": "BMI (per 5 kg/m²)"}
for col in NL_NUM:
    coefs[col] = coefs[col] / sd[col] * unit[col]
coefs = coefs.rename(labels)

nl_odds = pd.DataFrame({"Coefficient (log-odds)": coefs, "Odds ratio": np.exp(coefs)})
nl_odds.index.name = "Non-lab Logistic Regression (unweighted)"
reference_note = "Reference levels: gender = Female, smoking_status = current"
print(reference_note)
nl_odds.sort_values("Odds ratio", ascending=False).style.format("{:.3f}")
```

**📝 Interpretation:** all odds ratios are adjusted for the other predictors.
- **Hypertension (OR 2.1) and heart disease (OR 2.0)** each roughly double the odds of diabetes. These are the cardiometabolic comorbidities that screening guidelines already use as reasons to test.
- **Age (OR 1.6 per 10 years)** and **BMI (OR 1.6 per 5 kg/m²)** have similar, strong effects. Someone 30 years older has about 4.3 times the odds (1.62³), and moving from BMI 25 to 35 about 2.4 times (1.56²). **Male gender** raises the odds by about 40%.
- **Smoking:** compared with current smokers, never and past smokers have slightly lower odds (OR 0.83–0.85), and "unknown" much lower (OR 0.51). The "unknown" effect most likely reflects **who** has missing smoking data, not a protective effect. For example, 73% of patients under 18 have "No Info" against about 25% of adults, and a linear age term cannot fully absorb the very low diabetes rate in children. In cross-sectional data these smoking estimates should not be read causally.
- **`bmi_imputed`** has no effect once the other predictors are included (OR 1.01).

### <b>III <span style='color:#16C2D5'>|</span> Calibration and threshold</b>

Both models are calibrated with sigmoid and isotonic calibration (inner 3-fold CV) inside the outer 5-fold CV, as in section 5. The combination with the lowest out-of-fold Brier score becomes the final non-laboratory model. Thresholds are chosen from its calibrated OOF probabilities on the training set, for target recalls of 0.95, 0.90 and 0.85 and for the best F1, then applied **once** to the test set. The same thresholds for the full model (section 5) are shown for comparison.

```python
nl_oof = {}
for name, pipe in nl_pipelines.items():
    for method in ["sigmoid", "isotonic"]:
        est = CalibratedClassifierCV(clone(pipe), method=method, cv=inner_cv)
        nl_oof[(name, method)] = cross_val_predict(est, Xn_tr, y_tr, cv=cv, method="predict_proba")[:, 1]

nl_calib_table = pd.DataFrame({
    f"{name} + {method}": {"Brier score": brier_score_loss(y_tr, p), "Mean predicted risk": p.mean(),
                           "PR-AUC": average_precision_score(y_tr, p), "ROC-AUC": roc_auc_score(y_tr, p)}
    for (name, method), p in nl_oof.items()}).T
nl_calib_table.index.name = f"Non-lab, training-set OOF (prevalence {y_tr.mean():.3f})"

nl_best_name, nl_best_method = min(nl_oof, key=lambda k: brier_score_loss(y_tr, nl_oof[k]))
print(f"Chosen non-laboratory model: {nl_best_name} + {nl_best_method} calibration")
nl_calib_table.style.format("{:.4f}")
```

```python
# Final non-laboratory model, calibrated on the full training set; test set touched once
nl_calibrated = CalibratedClassifierCV(clone(nl_pipelines[nl_best_name]), method=nl_best_method, cv=cv)
nl_calibrated.fit(Xn_tr, y_tr)
p_te_nl = nl_calibrated.predict_proba(Xn_te)[:, 1]
p_oof_nl = nl_oof[(nl_best_name, nl_best_method)]

# Thresholds chosen on the training-set OOF probabilities (same rules as section 5)
prec_oof_nl, rec_oof_nl, thr_oof_nl = precision_recall_curve(y_tr, p_oof_nl)
prec_oof_nl, rec_oof_nl = prec_oof_nl[:-1], rec_oof_nl[:-1]
f1_oof_nl = 2 * prec_oof_nl * rec_oof_nl / np.clip(prec_oof_nl + rec_oof_nl, 1e-12, None)
nl_candidates = {f"Recall ≈ {r:.2f}": thr_oof_nl[np.where(rec_oof_nl >= r)[0].max()] for r in (0.95, 0.90, 0.85)}
nl_candidates["Best F1"] = thr_oof_nl[np.argmax(f1_oof_nl)]

nl_points = {}
for k, t in nl_candidates.items():
    nl_points[f"Non-lab: {k}"] = operating_point(y_te, p_te_nl, t)
    nl_points[f"Full (sec. 5): {k}"] = thr_test.loc[k].to_dict()
nl_thr_test = pd.DataFrame(nl_points).T
nl_thr_test[COUNT_COLS] = nl_thr_test[COUNT_COLS].astype(int)
nl_thr_test["NPV"] = nl_thr_test["TN"] / (nl_thr_test["TN"] + nl_thr_test["FN"])
nl_thr_test.index.name = "Held-out test set (thresholds from training OOF)"

print(f"Test set: Brier {brier_score_loss(y_te, p_te_nl):.4f} | mean predicted risk {p_te_nl.mean():.3f} "
      f"(observed {y_te.mean():.3f}) | PR-AUC {average_precision_score(y_te, p_te_nl):.3f} | "
      f"ROC-AUC {roc_auc_score(y_te, p_te_nl):.3f}")
nl_thr_test.style.format({**METRIC_FMT, "NPV": "{:.3f}"})
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5))
curves = {"Non-laboratory model": (p_te_nl, "#eb6834"), "Full model, with labs (section 5)": (p_te_cal, "#2a78d6")}

for name, (proba, color) in curves.items():
    fpr, tpr, _ = roc_curve(y_te, proba)
    axes[0].plot(fpr, tpr, color=color, lw=2, label=f"{name} (AUC {roc_auc_score(y_te, proba):.3f})")
    prec, rec, _ = precision_recall_curve(y_te, proba)
    axes[1].plot(rec, prec, color=color, lw=2, label=f"{name} (AP {average_precision_score(y_te, proba):.3f})")
for color, (k, t) in zip(["#1baf7a", "#eda100", "#533483", "#52514e"], nl_candidates.items()):
    row = nl_thr_test.loc[f"Non-lab: {k}"]
    axes[1].plot(row["Recall"], row["Precision"], "o", ms=9, color=color, markeredgecolor="white",
                 markeredgewidth=1.5, label=f"Non-lab {k}: t = {t:.3f}")

axes[0].plot([0, 1], [0, 1], color=MUTED, ls="--", lw=1)
axes[0].set(title="ROC curves (test set)", xlabel="False positive rate (1 − specificity)",
            ylabel="True positive rate (recall)")
axes[1].axhline(y_te.mean(), color=MUTED, ls="--", lw=1)
axes[1].set(title="Precision–recall curves (test set)", xlabel="Recall (sensitivity)",
            ylabel="Precision (PPV)", ylim=(0, 1.02))

# Calibration of the non-lab model on the test set (quantile bins: risks are spread out, not 0/1)
frac_pos, mean_pred = calibration_curve(y_te, p_te_nl, n_bins=10, strategy="quantile")
axes[2].plot([0, 0.6], [0, 0.6], color=MUTED, ls="--", lw=1, label="Perfect calibration")
axes[2].plot(mean_pred, frac_pos, "-o", color="#eb6834", lw=2, ms=6, markeredgecolor="white",
             label=f"Non-laboratory model (Brier {brier_score_loss(y_te, p_te_nl):.4f})")
axes[2].set(title="Calibration of the non-lab model (test set, deciles)", xlabel="Mean predicted risk",
            ylabel="Observed fraction diabetic")

for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)
    ax.grid(True, color="#e6e5e1", linewidth=0.8)
axes[0].legend(frameon=False, loc="lower right", fontsize=9)
axes[1].legend(frameon=False, loc="center left", bbox_to_anchor=(0.04, 0.72), fontsize=8)
axes[2].legend(frameon=False, loc="upper left", fontsize=9)
plt.tight_layout()
plt.show()
```

**📝 Interpretation and recommendation:**
- **Calibration:** all four combinations calibrate equally well (OOF Brier 0.068–0.069). Isotonic XGBoost has the lowest Brier score and is chosen. On the test set its mean predicted risk (9.0%) matches the observed prevalence (8.8%), and the decile curve lies on the diagonal. Unlike the lab model, the predicted risks are **spread out** (top decile about 35%), not pushed to 0 or 1. This is how a real risk score behaves.
- **The thresholds transfer to the test set:** target recalls of 0.95, 0.90 and 0.85 give test recalls of 0.948, 0.884 and 0.841.
- **What a pre-test triage rule would look like (threshold 0.028, recall ≈ 0.95):** 57% of patients are sent for an HbA1c or glucose test, and they include **95% of diabetics** (89 missed of 1,696). The other **43% are spared a blood test**, with an NPV of 98.9%. That is about 6.9 tests per case found, against 11.3 if everyone were tested (1 / 0.088). At threshold 0.056 (recall ≈ 0.88), 45% are tested and 196 diabetics are missed.
- **Comparison with the full model:** the full model looks far better at every threshold (for example, 2,253 against 9,404 false alarms at recall ≈ 0.95). That comparison is unfair to the non-laboratory model, because the full model **needs the very test the triage step is meant to ration**. The two models answer different questions. The non-laboratory model decides **who to test**, and the lab-based model (or simply the ADA/WHO criteria) interprets the result.
- **Recommendation:** for pre-test screening, use the calibrated non-laboratory model with a **low threshold (about 0.03 to 0.06)**, chosen according to how many missed cases are acceptable. The non-laboratory Logistic Regression is a near-equivalent, fully transparent alternative that could be turned into a points score. Its performance on this dataset should be confirmed on real data, because the dataset's labels are rule-generated (section 7).

# <span style="color:#E888BB; font-size: 1%;">7 | LIMITATIONS</span>
<div style="padding: 35px;color:white;margin:10;font-size:170%;text-align:left;display:fill;border-radius:10px;overflow:hidden;background:linear-gradient(90deg, #1a1a2e 0%, #16213e 40%, #0f3460 70%, #533483 100%);"><b>7 | LIMITATIONS</b></div>

1. **The outcome labels look rule-generated.** HbA1c ≥ 6.8% or glucose ≥ 220 mg/dL means *always* diabetic, HbA1c ≤ 5.0% or glucose ≤ 100 mg/dL means *never* diabetic, and in between the rate is flat at about 8.5% (EDA). The SHAP step functions confirm that the model learnt these cut-offs. The very high scores (ROC-AUC 0.98, precision 1.0 up to about 70% recall) measure how well the trees recover this rule, **not** expected clinical performance. On real data, where HbA1c and glucose relate to diagnosis continuously and lab values are noisy, performance would be clearly lower. The dataset's provenance is not documented, and it may be partly or fully synthetic.
2. **Circularity (incorporation bias).** HbA1c ≥ 6.5% and fasting glucose ≥ 126 mg/dL (or random glucose ≥ 200 mg/dL) are themselves diagnostic criteria (ADA/WHO). A model that uses them partly predicts the diagnosis from the test that defines it. It is also unknown whether glucose was measured fasting or random. For true pre-test screening ("who should be tested?"), the relevant model uses only non-laboratory predictors. Section 6 builds it: without HbA1c and glucose, ROC-AUC falls from 0.98 to 0.83 and PR-AUC from 0.88 to 0.31. Those figures are a more honest estimate of what can be predicted before testing, although they still come from the same, possibly synthetic, dataset.
3. **BMI imputation.** About 22.5% of raw rows have BMI = 27.32, the exact column mean. These values are mean-imputed, not measured. Mean imputation shrinks BMI's variance and biases its estimated effect (for example, the Logistic Regression odds ratio) towards zero. The imputed patients also differ systematically from the others (section 1), so the data are probably not missing at random. The sensitivity analysis shows little effect on discrimination, but with the original missingness pattern, multiple imputation would be the better approach.
4. **Duplicates.** 3,854 exact duplicate rows were dropped. Most of them share the imputed BMI and are probably distinct patients who collide by chance. Keeping them makes no difference to performance (sensitivity analysis), but it shifts the prevalence slightly (8.82% → 8.50%).
5. **Other data-quality issues.** Smoking status is "No Info" for 34% of patients. Age is capped at 80 and includes young children, and type 1 and type 2 diabetes are not distinguished. A few BMI values are implausible (> 80). HbA1c and glucose take only 18 distinct values each.
6. **Missing predictors and design.** The data has no ethnicity, family history, physical activity, diet, waist circumference or medication. It is cross-sectional, so the model detects existing diabetes and cannot predict future onset.
7. **Calibration and threshold are internal only.** The probabilities are calibrated to this dataset's 8.8% prevalence. Applying the model to a population with a different prevalence needs recalibration and a re-chosen threshold. The test set comes from the same source, so there is **no external validation**.
8. **Modelling choices.** The hyperparameter grids were small, and the sensitivity analyses reuse the tuned settings without re-tuning. SHAP values explain the class-weighted model on the log-odds scale, not the calibrated probabilities. Because calibration is monotone, the feature ranking is unaffected.
