🎯 Project Objective

Diabetes is one of the most prevalent chronic diseases worldwide, placing a substantial burden on individuals and healthcare systems. Early identification of individuals at risk can support timely intervention and improve long-term health outcomes.

The aim of this project is to develop a machine learning model capable of accurately classifying diabetes status using routinely collected demographic and clinical characteristics. The analysis explores the relationships between key risk factors, including **age**, **gender**, **body mass index (BMI)**, **hypertension**, **heart disease**, **smoking history**, **HbA1c level**, and **blood glucose level**.

From a clinician-data analyst perspective, understanding these variables is important not only for prediction but also for gaining insights into the factors associated with diabetes. Through exploratory data analysis, feature assessment, and predictive modelling, this project investigates how patient characteristics contribute to diabetes risk and evaluates the ability of machine learning algorithms to support clinical decision-making.

The findings may contribute to a better understanding of diabetes risk profiles and demonstrate the potential of data-driven approaches in preventive healthcare and population health management.

### <b>I <span style='color:#16C2D5'>|</span> Data preparation</b> 

```python
# Define a function to map the existing categories to new ones
def recategorize_smoking(smoking_status):
    if smoking_status in ['never', 'No Info']:
        return 'non-smoker'
    elif smoking_status == 'current':
        return 'current'
    elif smoking_status in ['ever', 'former', 'not current']:
        return 'past_smoker'

# Apply the function to the 'smoking_history' column
df['smoking_history'] = df['smoking_history'].apply(recategorize_smoking)

# Check the new value counts
print(df['smoking_history'].value_counts())
```

```python
df_encoded = df.copy()
```

### <b>II <span style='color:#16C2D5'>|</span> Encoding</b> 

```python
def perform_one_hot_encoding(df, column_name):
    # Perform one-hot encoding on the specified column
    dummies = pd.get_dummies(df[column_name], prefix=column_name)

    # Drop the original column and append the new dummy columns to the dataframe
    df = pd.concat([df.drop(column_name, axis=1), dummies], axis=1)

    return df

# Perform one-hot encoding on the gender variable
df_encoded = perform_one_hot_encoding(df_encoded, 'gender')

# Perform one-hot encoding on the smoking history variable
df_encoded = perform_one_hot_encoding(df_encoded, 'smoking_history')
```

## Logistic Regression Performance

The logistic regression model achieved an overall accuracy of 95.9% on the test dataset. The model showed excellent performance in identifying non-diabetic individuals (recall = 0.99). For diabetic individuals, precision was 0.86, indicating that most positive predictions were correct. However, recall was 0.61, suggesting that approximately 39% of diabetic individuals were not identified by the model. Therefore, while the model demonstrated strong overall predictive performance, its sensitivity for detecting diabetes could be improved.
